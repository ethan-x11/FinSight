```python
testlist =[{"key1": "value1"}, {"key2": "value2"}, {"key3": "value3"}, {"key4": "value4"}, {"key2": "value2"}]

testset = set()
for item in testlist:
    testset.add(frozenset(item.items()))

finallist = [dict(items) for items in testset]

print(finallist)
```
